```python
import numpy as np

arr=[32,34,12,42,3,62,46,23,5,1,54,23,353,69,76,455,12]
minimum,q1,median,q3,maximum = np.quantile(arr,[0,0.25,0.5,0.75,1])

minimum,q1
maximum
```

```python
IQR = q3-q1
IQR
```

```python
#handling outlier 
lower_fence = q1-1.5*(IQR)
higher_fence = q3+ 1.5*(IQR)
print(higher_fence,lower_fence)
#any value higher than higher_fence and lower then lower_fence is considered as outlier 
```
