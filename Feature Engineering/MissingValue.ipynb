```python
#Column wise deletiom
df.dropna(axis=1)
```

```python
df['age']
```

```python
#handling missing values with mean imputation
df['Age-Mean']=df['age'].fillna(df['age'].mean())
df[['Age-Mean','age']]
```

```python
#Handling median imputation
df['Median-Age']=df['age'].fillna(df['age'].median())
df[['age','Age-Mean','Median-Age']]
```

```python
#mode imputation only for categorical values 
mode_value = df[df['embarked'].notna()].mode()[0]          # use to find out the most used median to fill at the na places
```

```python
df['embarked']=df['embarked'].fillna(mode_value)
```
