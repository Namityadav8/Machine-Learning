```python
import matplotlib.pyplot as plt


x=[1,2,3,4,5]
y=[1,4,9,16,25]


plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Graph')
```

Customizing the line graph

```python

plt.plot(x,y,color='blue',linestyle=':')
```

```python
plt.plot(x,y,color='red',linestyle='--',marker='*',markersize=10,linewidth=4)
plt.grid(True)
```
