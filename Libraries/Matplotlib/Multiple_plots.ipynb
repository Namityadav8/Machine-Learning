#Creating multiple plots 

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
x=[1,2,3,4,5]
y=[5,6,7,8,9]
plt.subplot(2,3,1)
plt.title("plot-1")
plt.plot(x,y,color='red',marker='*',markersize=10,linestyle='--')


plt.subplot(1,3,2)
plt.title("plot-2")
plt.plot(x,y,color='blue',marker='o',markersize=12,linestyle='-')
```
