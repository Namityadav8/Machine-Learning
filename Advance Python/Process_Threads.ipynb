### Multithreading

```python
import threading
import time
```

```python
def print_time():
    for i in range(5):
        time.sleep(2)
        print(i)

def print_letter():
    for letter in "abcde":
        time.sleep(2)
        print(letter)
```

```python
#Create two threads as follows  for multithreading to run concurrently
t1=threading.Thread(target=print_time)
t2=threading.Thread(target=print_letter)

t=time.time()   


#starting the two threads
t1.start()
t2.start()


#waiting for the threads to complete 
t1.join()
t2.join()

## using this process we can run two functions concurrently that will remove the time delay of the first function
finished_time=time.time()-t
print(finished_time)
```

```python
t=time.time()
print_time()
print_letter()
f_time = time.time()-t
print(f_time)
```
