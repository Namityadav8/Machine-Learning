```python
import sqlite3
```

```python
#connect to an SQLite database
connect = sqlite3.connect('test.db')
connect  #will create an test.db file in the current directory
```

```python
#creating cursor to execute SQL commands
cursor = connect.cursor()
cursor
```

```python
#execute SQL commands using "execute" method

#creating a table
cursor.execute('''
               CREATE TABLE IF NOT EXISTS stocks(
               id INTEGER PRIMARY KEY,
               name TEXT NOT NULL,
               age INTEGER NOT NULL,
               department TEXT NOT NULL
               )
               ''')




#commit the change to get it executed

connect.commit()
```

```python
#fetching the data to check if the table is created

cursor.execute('SELECT * FROM stocks')
```

```python
#inserting into the table
cursor.execute('''
               INSERT INTO stocks (name, age, department)
               VALUES ('John', 25, 'HR')
               ''')

connect.commit()
```

```python
#Querying the data

cursor.execute('SELECT * FROM stocks')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
 #Updating the data
cursor.execute('''UPDATE stocks SET age=26 WHERE name='John' ''')
connect.commit()
```
