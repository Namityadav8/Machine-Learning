### logging with multiple loggers

```python
import logging
```

```python
log1=logging.getLogger('module_1')
log2=logging.getLogger('module_2')
log1.setLevel(logging.WARNING)
log2.setLevel(logging.DEBUG)
```

```python
logging.basicConfig(
    filemode='w',
    filename='module_1.log',
    level=logging.DEBUG,
)
logging.basicConfig(
    filemode='w',
    filename='module_2.log',
    level=logging.DEBUG,
)
```

```python
log1.warning('This is a warning message')
log2.debug('This is a debug message')
```
